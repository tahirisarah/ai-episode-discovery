```python
def run_search(btn):
  with output_area:
    clear_output()
    selected_show = show_dropdown.value.strip()
    char_query = character_input.value.strip().lower()
    event_query = event_input.value.strip().lower()
    object_query = object_input.value.strip().lower()
    location_query = location_input.value.strip().lower()
    descriptor_query = descriptor_input.value.strip().lower()
    actor_query = actor_input.value.strip().lower()

    #  I need to check that at least one field is filled
    if not (char_query or event_query or object_query or location_query or descriptor_query or actor_query):
      print("Please enter at least one keyword to search")
      return

    # System need to Filter the episodes to the selected show only
    show_episodes = episodes_full[episodes_full["show_name"] == selected_show].copy()

    if len(show_episodes) == 0:
      print(f"No episodes found for {selected_show} in the database")
      return

    # AND Logic in which each filled field must match

    matched_episode_ids = set(show_episodes["episode_id"].tolist())

    # For the character field, it must appear in ep_chars joined with the characters table
    if char_query:
      chars_merged = ep_chars.merge(characters[["character_id","first_name","last_name"]], on="character_id")
      chars_merged["full_name"] = chars_merged["first_name"].str.lower() + " " + chars_merged["last_name"].str.lower()
      char_match = chars_merged[
          chars_merged["first_name"].str.lower().str.contains(char_query) |
          chars_merged["last_name"].str.lower().str.contains(char_query) |
          chars_merged["full_name"].str.lower().str.contains(char_query)
      ]

      matched_episode_ids = matched_episode_ids & set(char_match["episode_id"].tolist())

    # For the actor field, it must appear in characters table as actor_name
    if actor_query:
      actor_match = characters[characters["actor_name"].str.contains(actor_query, case=False, na=False)]
      actor_episodes = ep_chars[ep_chars["character_id"].isin(actor_match["character_id"])]["episode_id"]
      matched_episode_ids = matched_episode_ids & set(actor_episodes.tolist())

    # For the event field, it must match a keyword row with cateory = event
    if event_query:
      event_keywords = keywords[keywords["keyword_category"].str.lower().str.contains("event")]
      event_match = event_keywords[event_keywords["keyword_name"].str.lower().str.contains(event_query)]
      event_episodes = ep_keys[ep_keys["keyword_id"].isin(event_match["keyword_id"])]["episode_id"]
      matched_episode_ids = matched_episode_ids & set(event_episodes.tolist())

    # For the object field, it must match a keyword row with cateory = event
    if object_query:
      object_keywords = keywords[keywords["keyword_category"].str.lower().str.contains("object")]
      object_match = object_keywords[object_keywords["keyword_name"].str.lower().str.contains(object_query)]
      object_episodes = ep_keys[ep_keys["keyword_id"].isin(object_match["keyword_id"])]["episode_id"]
      matched_episode_ids = matched_episode_ids & set(object_episodes.tolist())

    # For the event field, it must match a keyword row with cateory = event
    if location_query:
      location_keywords = keywords[keywords["keyword_category"].str.lower().str.contains("place")]
      location_match = location_keywords[location_keywords["keyword_name"].str.lower().str.contains(location_query)]
      location_episodes = ep_keys[ep_keys["keyword_id"].isin(location_match["keyword_id"])]["episode_id"]
      matched_episode_ids = matched_episode_ids & set(location_episodes.tolist())

    # For the descriptor field, it must match a keyword row with cateory = event
    if descriptor_query:
      descriptor_keywords = keywords[keywords["keyword_category"].str.lower().str.contains("descriptor")]
      descriptor_match = descriptor_keywords[descriptor_keywords["keyword_name"].str.lower().str.contains(descriptor_query)]
      descriptor_episodes = ep_keys[ep_keys["keyword_id"].isin(descriptor_match["keyword_id"])]["episode_id"]
      matched_episode_ids = matched_episode_ids & set(descriptor_episodes.tolist())

    # If nothing matches, it will prompt the user to enter different keywords
    if len(matched_episode_ids) == 0:
      print("No episodes matched your search. Please enter valid keywords")
      return

    results = show_episodes[show_episodes["episode_id"].isin(matched_episode_ids)]
    results = results.sort_values("episode_number")

    print(f"{len(results)} episode(s) found in {selected_show}")

    for _,row in results.iterrows():
      print(f"{row['episode_number']}. {row['title']}")
      print(f"Summary {row['plot_summary']}")
```
